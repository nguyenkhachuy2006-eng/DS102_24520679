```python
import numpy as np
```

```python
def newton_raphson(grad, hess, x0, tol=1e-8, max_iter=100):
    x = np.array(x0, dtype=float)
    for k in range(max_iter):
        g = grad(x)
        if np.linalg.norm(g) < tol:
            break
        d = np.linalg.solve(hess(x), g)
        x = x - d
    return x, k
```

**Ví dụ**: f(x, y) = (x - 1)^2 + 2*(y + 2)^2 + x*y

```python
def f(x):
    return (x[0] - 1)**2 + 2*(x[1] + 2)**2 + x[0]*x[1]

def grad(x):
    return np.array([
        2*(x[0] - 1) + x[1],
        4*(x[1] + 2) + x[0]
    ])

def hess(x):
    return np.array([
        [2.0, 1.0],
        [1.0, 4.0]
    ])
```

```python
x_opt, iters = newton_raphson(grad, hess, x0=[0, 0])
print("Điểm cực trị:", x_opt)
print("f tại điểm đó:", f(x_opt))
print("Số vòng lặp:", iters)
```

```python
eig = np.linalg.eigvalsh(hess(x_opt))
if np.all(eig > 0):
    print("Cực tiểu")
elif np.all(eig < 0):
    print("Cực đại")
else:
    print("Điểm yên ngựa / không xác định")
```
